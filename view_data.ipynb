```python
import numpy as np
import matplotlib.pyplot as plt

# Veriyi yükle
seismic = np.load('data/train/train_seismic.npy')
labels = np.load('data/train/train_labels.npy')

print('Seismic shape:', seismic.shape)
print('Labels shape:', labels.shape)
print('Label sınıfları:', np.unique(labels))
```

```python
# Sismik kesit görüntüle (inline=100. kesit)
kesit = 100

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].imshow(seismic[kesit, :, :].T, cmap='seismic', aspect='auto')
axes[0].set_title(f'Sismik - inline {kesit}')
axes[0].set_xlabel('crossline')
axes[0].set_ylabel('depth')

axes[1].imshow(labels[kesit, :, :].T, cmap='tab10', aspect='auto')
axes[1].set_title(f'Etiketler - inline {kesit}')
axes[1].set_xlabel('crossline')
axes[1].set_ylabel('depth')

plt.tight_layout()
plt.show()
```

```python
# Farklı yönlerden kesit (crossline)
kesit = 350

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].imshow(seismic[:, kesit, :].T, cmap='seismic', aspect='auto')
axes[0].set_title(f'Sismik - crossline {kesit}')

axes[1].imshow(labels[:, kesit, :].T, cmap='tab10', aspect='auto')
axes[1].set_title(f'Etiketler - crossline {kesit}')

plt.tight_layout()
plt.show()
```
